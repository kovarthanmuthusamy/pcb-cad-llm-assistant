```python
from smolvlm import smolvlm_inference
from qwen25_inference import qwen_inference
from granite_vision32_inference import granite_inference
from llava_1_6_mistral_inference import llava_inference
import time
```

### Circuit Image Description Inference

```python
image = "circuit_imgs_03/P2P_R210_L75_CHARIMP65_PROP0.2.png" 

#-----------------LLAVA-1.6-Mistral-7B Model-------------------------#

llava_start_time = time.time()
llava_output = llava_inference(image, adapter=False)
llava_end_time = time.time()

llava_inference_time = llava_end_time - llava_start_time

llava_adapter_start_time = time.time()
llava_output_adapter = llava_inference(image, adapter=True)
llava_adapter_end_time = time.time()

llava_adapter_inference_time = llava_adapter_end_time - llava_adapter_start_time

#-----------------SmolVLM Intruct Model------------------------------#
smolvlm_start_time = time.time()
smolvlm_output = smolvlm_inference(image, model="1", adapter=False)
smolvlm_end_time = time.time() 

smolvlm_inference_time = smolvlm_end_time - smolvlm_start_time

smolvlm_adapter_start_time = time.time()
smolvlm_output_adapter = smolvlm_inference(image, model="1", adapter=True)
smolvlm_adapter_end_time = time.time()

smolvlm_adapter_inference_time = smolvlm_adapter_end_time - smolvlm_adapter_start_time

#-----------------SmolVLM 500M Intruct Model--------------------------#

smolvlm_500m_start_time = time.time()
smolvlm_500m_output = smolvlm_inference(image, model="2", adapter=False)
smolvlm_500m_end_time = time.time() 

smolvlm_500m_inference_time = smolvlm_500m_end_time - smolvlm_500m_start_time

smolvlm_500m_adapter_start_time = time.time()
smolvlm_500m_output_adapter = smolvlm_inference(image, model="2", adapter=True)
smolvlm_500m_adapter_end_time = time.time()

smolvlm_500m_adapter_inference_time = smolvlm_500m_adapter_end_time - smolvlm_500m_adapter_start_time

#-----------------Qwen 2.5 VL Intruct Model----------------------------#

qwen_start_time = time.time()
qwen_output = qwen_inference(image, adapter=False)
qwen_end_time = time.time()

qwen_inference_time = qwen_end_time - qwen_start_time

qwen_adapter_start_time = time.time()
qwen_output_adapter = qwen_inference(image, adapter=True)
qwen_adapter_end_time = time.time()

qwen_adapter_inference_time = qwen_adapter_end_time - qwen_adapter_start_time

#-----------------Granite Vision 3.2 Model------------------------------#

granite_start_time = time.time()
granite_output = granite_inference(image, adapter=False)
granite_end_time = time.time()

granite_inference_time = granite_end_time - granite_start_time

granite_adapter_start_time = time.time()
granite_output_adapter = granite_inference(image, adapter=True)
granite_adapter_end_time = time.time()

granite_adapter_inference_time = granite_adapter_end_time - granite_adapter_start_time
```

```python
print("LlaVA-1.6-Mistral output without adapter:\n")
print(llava_output)
print("Inference Time: ",llava_inference_time)

print("\nLlaVA-1.6-Mistral output with adapter:\n")
print(llava_output_adapter)
print("Inference Time: ",llava_adapter_inference_time)

print("\nSMOLVLM output without adapter:\n")
print(smolvlm_output)
print("Inference Time: ",smolvlm_inference_time)

print("\nSMOLVLM output with adapter:\n")
print(smolvlm_output_adapter)
print("Inference Time: ",smolvlm_adapter_inference_time)

print("\nSMOLVLM 500M output without adapter:\n")
print(smolvlm_500m_output)
print("Inference Time: ",smolvlm_500m_inference_time)

print("\nSMOLVLM output with adapter:\n")
print(smolvlm_500m_output_adapter)
print("Inference Time: ",smolvlm_500m_adapter_inference_time)

print("\nQWEN 2.5 output without adapter:\n")
print(qwen_output)
print("Inference Time: ",qwen_inference_time)

print("\nQWEN 2.5 output with adapter:\n")
print(qwen_output_adapter)
print("Inference Time: ",qwen_adapter_inference_time)

print("\nGranite 3.2 output without adapter:\n")
print(granite_output)
print("Inference Time: ",granite_inference_time)

print("\nGranite 3.2 output with adapter:\n")
print(granite_output_adapter)
print("Inference Time: ",granite_adapter_inference_time)
```
